BAGIAN 4 (A)

```python
# Import library yang diperlukan
import pandas as pd
from scipy.stats import ttest_ind

# Load data Titanic
file_path = 'train.csv'
data = pd.read_csv(file_path)

# Menampilkan beberapa baris data untuk memastikan formatnya
data.head()

# Menghapus data yang memiliki nilai Age yang hilang
data = data.dropna(subset=['Age'])

# Memisahkan data berdasarkan status kelangsungan hidup
survived = data[data['Survived'] == 1]['Age']
not_survived = data[data['Survived'] == 0]['Age']

# Melakukan uji t independen
t_stat, p_value = ttest_ind(survived, not_survived, equal_var=False)

# Menampilkan hasil uji t
print("Hasil Uji T:")
print(f"T-Statistic: {t_stat:.4f}")
print(f"P-Value: {p_value:.4f}")
```

1. Hasil Uji T:
    - T-Statistic: -2.0460
    - P-Value: 0.0412

2. Hipotesis:
    - H0 (Null Hypothesis): Tidak terdapat perbedaan usia yang signifikan antara penumpang yang selamat dan tidak selamat.
    - H1 (Alternative Hypothesis): Terdapat perbedaan usia yang signifikan antara penumpang yang selamat dan tidak selamat.

3. Keputusan:
    - Dengan tingkat signifikansi (𝛼) = 0.05:
    - P-value = 0.0412 < 0.05

4. Kesimpulan:
Terdapat perbedaan usia yang signifikan secara statistik antara penumpang yang selamat dan tidak selamat pada kapal Titanic.
menolak Null Hypothesis karena ada bukti statistik bahwa usia penumpang yang selamat berbeda secara signifikan dari usia penumpang yang tidak selamat.

BAGIAN 4 (B)

```python
from scipy.stats import chi2_contingency

# Membuat tabel kontingensi berdasarkan jenis kelamin dan kelangsungan hidup
contingency_table = pd.crosstab(data['Sex'], data['Survived'])

# Menampilkan tabel kontingensi
print("Tabel Kontingensi:")
print(contingency_table)

# Melakukan uji chi-square
chi2_stat, p_value, dof, expected = chi2_contingency(contingency_table)

# Menampilkan hasil uji chi-square
print("\nHasil Uji Chi-Square:")
print(f"Chi-Square Statistic: {chi2_stat:.4f}")
print(f"P-Value: {p_value:.4e}")
print(f"Degrees of Freedom: {dof}")
print("\nExpected Frequencies:")
print(expected)
```

1. HASIL KEPUTUSAN
- menolak null hipotesis (H0), yang menyatakan bahwa tidak terdapat hubungan signifikan antara jenis kelamin dan kelangsungan hidup. Sebaliknya, data mendukung hipotesis alternatif (H1), yaitu terdapat hubungan yang signifikan antara jenis kelamin dan kelangsungan hidup.
