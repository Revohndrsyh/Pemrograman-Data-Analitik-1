```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
world_happy = pd.read_csv('world_happiness.csv')
world_happy.head()
```

```python
# Plot scatter antara happiness_score dan life_exp
plt.figure(figsize=(10, 6))
plt.scatter(world_happy['life_exp'], world_happy['happiness_score'], color='b', alpha=0.5)
plt.title("Hubungan antara Skor Kebahagiaan dan Harapan Hidup")
plt.xlabel("Life Exp")
plt.ylabel("Happiness Score")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(6, 6))
plt.scatter(world_happy['life_exp'], world_happy['happiness_score'], color='b', alpha=0.5)

# Menghitung Regresi Linear untuk trendline
m, b = np.polyfit(world_happy['life_exp'], world_happy['happiness_score'],1)
plt.plot(world_happy['life_exp'],m * world_happy['life_exp'] + b, color='blue', linewidth=2, label='trendline')

plt.title("Hubungan antara Skor Kebahagiaan dan Harapan Hidup")
plt.xlabel("Life Exp")
plt.ylabel("Happiness Score")
plt.legend()
plt.grid(True)
plt.show()
```

```python
Correlation = world_happy['life_exp']. corr(world_happy['happiness_score'])
print("Korelasi antara life exp dan Happiness score : ", Correlation)
```

```python
# sccater plot antara gdp_per_cap dan life_exp
plt.figure(figsize=(10, 6))
plt.scatter(world_happy['gdp_per_cap'], world_happy['life_exp'], color='red', alpha=0.5)
plt.title("Hubungan antara GDP per Capita dan Harapan Hidup")
plt.xlabel("GDP per Capita")
plt.ylabel("Life Expectancy")
plt.grid(True)
plt.show()

# Menghitung korelasi antara gdp_per_cap dan life_exp
correlation = world_happy['gdp_per_cap'].corr(world_happy['life_exp'])
print("Korelasi antara gdp_per_cap dan life_exp:", correlation)
```

```python
# Scatter plot antara gdp_per_cap dan happiness_score
plt.figure(figsize=(10, 6))
plt.scatter(world_happy['gdp_per_cap'], world_happy['happiness_score'], color='green', alpha=0.5)
plt.title("Hubungan antara GDP per Capita dan Happiness Score")
plt.xlabel("GDP per Capita")
plt.ylabel("Happiness Score")
plt.grid(True)
plt.show()

# Menghitung korelasi antara gdp_per_cap dan happiness_score
correlation = world_happy['gdp_per_cap'].corr(world_happy['happiness_score'])
print("Korelasi antara gdp_per_cap dan happiness_score:", correlation)
```

```python
# Membuat kolom baru dengan logaritma dari gdp_per_cap
world_happy['log_gdp_per_cap'] = np.log(world_happy['gdp_per_cap'])

# Scatter plot antara log_gdp_per_cap dan happiness_score
plt.figure(figsize=(10, 6))
plt.scatter(world_happy['log_gdp_per_cap'], world_happy['happiness_score'], color='orange', alpha=0.5)
plt.title("Hubungan antara Log GDP per Capita dan Happiness Score")
plt.xlabel("Log GDP per Capita")
plt.ylabel("Happiness Score")
plt.grid(True)
plt.show()

# Menghitung korelasi antara log_gdp_per_cap dan happiness_score
correlation = world_happy['log_gdp_per_cap'].corr(world_happy['happiness_score'])
print("Korelasi antara log_gdp_per_cap dan happiness_score:", correlation)
```
